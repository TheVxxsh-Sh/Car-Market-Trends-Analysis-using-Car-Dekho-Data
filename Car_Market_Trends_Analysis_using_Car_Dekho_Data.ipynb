# Dataset Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Car Dekho dataset
df = pd.read_csv('1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv')
print("Dataset loaded successfully!")
display(df.head())
```

# Dataset Shape & Structure Examination

```python
# Examine dimensions and column structure
print("--- Dataset Shape & Structure ---")
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
print("\nColumns List:")
print(df.columns.tolist())
```

# Data Types & Missing Values Handling

```python
# Examine data types and check for missing values
print("--- Data Types & Non-Null Counts ---")
print(df.info())

print("\n--- Missing Values Check ---")
print(df.isnull().sum())
```

# Statistical Summary & Outlier Investigation

```python
# Descriptive statistics
print("--- Statistical Summary ---")
display(df.describe())

# Investigate outliers in Selling Price and Kms Driven
for col in ['Selling_Price', 'Kms_Driven']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outlier_count = df[(df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))].shape[0]
    print(f"Outliers in {col}: {outlier_count}")
```

# Univariate & Bivariate Analysis Visualizations

```python
sns.set_theme(style="whitegrid")

# 1. Univariate Analysis: Distribution of Selling Price
plt.figure(figsize=(8, 4))
sns.histplot(df['Selling_Price'], kde=True, color='teal')
plt.title('Univariate Analysis: Distribution of Selling Price (Lakhs)')
plt.xlabel('Selling Price (Lakhs)')
plt.ylabel('Frequency')
plt.show()

# 2. Bivariate Analysis: Present Price vs Selling Price by Fuel Type
plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x='Present_Price', y='Selling_Price', hue='Fuel_Type', palette='deep', alpha=0.8)
plt.title('Bivariate Analysis: Present Price vs. Selling Price by Fuel Type')
plt.xlabel('Present Showroom Price (Lakhs)')
plt.ylabel('Selling Price (Lakhs)')
plt.show()
```

# Categorical Analysis & Average Selling Price by Fuel Type

```python
# Group by fuel type to compare pricing trends
fuel_summary = df.groupby('Fuel_Type')['Selling_Price'].mean().reset_index()
print("--- Average Selling Price by Fuel Type ---")
display(fuel_summary)

# Bar plot visualization
plt.figure(figsize=(8, 4))
sns.barplot(data=fuel_summary, x='Fuel_Type', y='Selling_Price', palette='viridis')
plt.title('Average Selling Price by Fuel Type (INR Lakhs)')
plt.xlabel('Fuel Type')
plt.ylabel('Mean Selling Price')
plt.show()
```
